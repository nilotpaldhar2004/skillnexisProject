```python
df=pd.read_csv(r"/kaggle/input/datasets/nilotpaldhar/data27092026/data.csv")
```

```python
df.head()
```

# 1. Fix Quotes & Convert Dates to datetime

```python
df['Date'] = df['Date'].astype(str).str.replace("'", "").str.strip()
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce', format='mixed')
```

```python
df.head()
```

# 2. Handle Missing Values (Calories)

```python
print(df.isnull().sum())
```

```python
median_cal = df['Calories'].median()
df['Calories'] = df['Calories'].fillna(median_cal)
```

# 3. Fix Outliers / Erroneous Data (Duration)

```python
print(df[df['Duration'] > 120])
```

```python
df.loc[df['Duration'] == 450, 'Duration'] = 45
```

# 4. Remove Duplicates

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```
